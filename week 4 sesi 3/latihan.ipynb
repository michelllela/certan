Baris y = lab.fit_transform(y) menjalankan dua langkah sekaligus. Bagian fit membuat mesin itu mempelajari kategori apa saja yang ada di y, yaitu setosa, versicolor, dan virginica, lalu mengurutkannya secara alfabetis dan memberi nomor: setosa = 0, versicolor = 1, virginica = 2. Bagian transform kemudian mengganti setiap nama spesies di y dengan nomornya. Hasilnya disimpan kembali ke variabel y, sehingga isi y yang lama (teks) tertimpa oleh versi angkanya.

Baris terakhir, y, hanya menampilkan isinya di layar. Outputnya berupa array berisi 150 angka, yaitu 50 angka 0, lalu 50 angka 1, lalu 50 angka 2, sesuai urutan data aslinya.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Kode ini membagi data menjadi dua kelompok: data latih untuk mengajari model, dan data uji untuk menguji seberapa pintar model setelah belajar. Kalau model diuji dengan data yang sama seperti yang dipakai belajar, hasilnya tidak jujur, seperti murid yang diberi soal ujian yang sudah pernah dibahas.

Baris from sklearn.model_selection import train_test_split memanggil fungsi pembagi data dari scikit-learn. Fungsi ini mengacak urutan data terlebih dulu, lalu memotongnya menjadi dua bagian sesuai proporsi yang kita tentukan.

Baris X_train, X_test = train_test_split(X, test_size=0.2, random_state=0) membagi fitur (X). Parameter test_size=0.2 artinya 20% data dijadikan data uji, sehingga sisanya, 80%, menjadi data latih. Dari 150 data, hasilnya 120 untuk latih (X_train) dan 30 untuk uji (X_test). Parameter random_state=0 mengunci pengacakan supaya hasil pembagiannya selalu sama setiap kali kode dijalankan. Tanpa itu, pembagiannya berubah-ubah, sehingga akurasinya pun bisa berbeda tiap kali dijalankan.

Baris berikutnya, y_train, y_test = train_test_split(y, test_size=0.2, random_state=0), melakukan hal yang sama untuk label (y), yaitu kunci jawabannya. Karena test_size dan random_state yang dipakai sama persis, pengacakan yang terjadi juga identik. Akibatnya, baris ke-1 di X_train tetap berpasangan dengan baris ke-1 di y_train, dan begitu juga untuk data uji. Kalau random_state-nya berbeda, soal dan kunci jawabannya bisa tertukar dan model akan belajar dengan data yang salah pasangan.

Setelah kode ini, akan ada empat variabel: X_train dan y_train untuk melatih model, serta X_test dan y_test untuk menguji model. Cara yang lebih ringkas sebenarnya adalah X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0), tapi versi di modul juga menghasilkan pembagian yang sama.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Kode ini membuat model Naïve Bayes lalu melatihnya memakai data latih, sehingga model belajar mengenali pola ukuran bunga untuk tiap spesies.

Baris from sklearn.naive_bayes import GaussianNB memanggil algoritma Gaussian Naïve Bayes dari scikit-learn. Pilihan "Gaussian" dipakai karena keempat fitur Iris berupa angka kontinu (panjang dan lebar dalam cm) yang diasumsikan berdistribusi normal.

Baris model = GaussianNB() membuat sebuah model kosong dan menyimpannya di variabel model. Pada tahap ini model belum tahu apa-apa, ibarat murid baru yang belum belajar.

Baris NB_model = model.fit(X_train, y_train) adalah proses belajarnya. Fungsi fit() menerima X_train (ukuran bunga) dan y_train (spesiesnya), lalu untuk setiap spesies model menghitung rata-rata dan standar deviasi dari masing-masing fitur, serta seberapa sering spesies itu muncul di data latih. Angka-angka inilah yang nanti dipakai dalam rumus Gaussian dan Teorema Bayes saat model menebak data baru. Misalnya, model mencatat bahwa petal width setosa rata-ratanya sangat kecil, sedangkan virginica jauh lebih besar.

Hasil dari fit() disimpan ke NB_model. Sebenarnya fit() mengembalikan objek model yang sama dengan model, jadi NB_model dan model mengarah ke model terlatih yang sama. Itu sebabnya nanti kamu bisa memanggil NB_model.predict(...) untuk menebak data uji maupun data baru.

```python
result = NB_model.predict(X_test)
result
```

Kode ini meminta model yang sudah dilatih untuk menebak spesies dari data uji, yaitu data yang belum pernah dilihat model saat belajar.

Baris result = NB_model.predict(X_test) memberikan X_test (30 baris ukuran bunga) ke model. Untuk setiap baris, model menghitung probabilitas bunga itu termasuk setosa, versicolor, atau virginica, memakai rata-rata, standar deviasi, dan frekuensi kelas yang dipelajari saat fit(). Kelas dengan probabilitas tertinggi dipilih sebagai tebakan. Hasilnya disimpan di variabel result. Perhatikan bahwa y_test (kunci jawaban) tidak diberikan ke model di tahap ini, karena model harus menebak sendiri.

Baris terakhir, result, hanya menampilkan isinya di layar. Outputnya berupa array berisi 30 angka, misalnya array([2, 1, 0, 2, 0, ...]). Angka-angka itu adalah tebakan model dalam bentuk label numerik: 0 untuk setosa, 1 untuk versicolor, dan 2 untuk virginica. Urutannya mengikuti urutan baris di X_test, jadi angka pertama adalah tebakan untuk baris pertama data uji, dan seterusnya.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Kode ini mengukur seberapa akurat tebakan model, yaitu dengan membandingkan hasil tebakan dengan kunci jawaban yang sebenarnya.

Baris from sklearn.metrics import accuracy_score memanggil fungsi penghitung akurasi dari scikit-learn. Akurasi adalah persentase tebakan yang benar dari seluruh data uji.

Misalnya, kalau 29 dari 30 tebakan benar, hasilnya 29/30 = 0,9667.

Urutan argumen di accuracy_score sebaiknya selalu kunci jawaban dulu (y_test), baru tebakan (result).
